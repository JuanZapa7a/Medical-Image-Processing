## 1. Introduction

Blood cancer, specifically `Acute Lymphoblastic Leukemia (ALL) (see NB1), is a critical area of research` due to its aggressive nature and impact on both pediatric and adult populations. `The classification of blood cancer cells into different stages and types is essential for accurate diagnosis and treatment planning`. Deep learning has emerged as a powerful tool in this domain, offering unprecedented accuracy and efficiency in classifying blood cancer cells based on microscopic images or other diagnostic data.



### Importance of Deep Learning in Blood Cancer Classification

Deep learning, particularly Convolutional Neural Networks `(CNNs), has revolutionized the field of medical image analysis, including the classification of blood cancer cells`. Traditional methods of cancer diagnosis involve manual inspection by pathologists, which can be time-consuming and subject to human error. `Deep learning models, trained on large datasets of labeled images, can automatically identify and classify different types of blood cancer cells with high accuracy, thus assisting in early diagnosis and personalized treatment plans`. This automated process not only speeds up the diagnostic process but also enhances the reliability of the results, which is crucial in life-threatening conditions like ALL.



### Acute Lymphoblastic Leukemia (ALL)

Acute Lymphoblastic Leukemia (ALL) is a type of cancer of the blood and bone marrow that affects white blood cells, specifically lymphocytes. It is characterized by the `overproduction of immature lymphoblasts (precursor cells to lymphocytes) which do not mature properly, leading to a decrease in the production of normal blood cells`. This condition progresses rapidly, making early diagnosis and treatment critical.

### Pre-B, Early Pre-B, and Pro-B Cells

**1. `Pro-B Cells`:**
   - **Stage of Development**: Pro-B cells represent the `earliest stage of B-cell` development. These cells arise from hematopoietic stem cells in the bone marrow.
   - **Key Characteristics**: During this stage, the `cell begins rearranging its heavy chain genes (immunoglobulin genes), specifically the DJ (diversity-joining) segments, a process crucial for the development of a functional B-cell receptor (BCR)`.
   - **Markers**: `Pro-B cells express surface markers like CD19 and CD10 but do not yet express the heavy chain of the BCR`.

**2. `Early Pre-B Cells`:**
   - **Stage of Development**: `Early Pre-B cells follow the Pro-B stage` and continue to mature in the bone marrow.
   - **Key Characteristics**: In this stage, the `V-DJ rearrangement of the heavy chain is complete`, and the cell begins to express the heavy chain in association with a surrogate light chain, `forming the Pre-B cell receptor (Pre-BCR)`.
   - **Markers**: In addition to CD19 and CD10, early Pre-B cells `start expressing the μ heavy chain of the BCR`.

**3. `Pre-B Cells`:**
   - **Stage of Development**: `Pre-B cells represent the stage after early Pre-B cells`.
   - **Key Characteristics**: These cells undergo `light chain rearrangement to form a complete and functional B-cell receptor (BCR)`. The successful expression of a functional BCR is crucial for the cell to progress to the immature B-cell stage.
   - **Markers**: `Pre-B cells express the complete BCR on their surface, along with markers like CD19, CD10, and CD20`.



### Relevance to Acute Lymphoblastic Leukemia

`In ALL, malignant transformation can occur at different stages of B-cell development, including Pro-B, Early Pre-B, and Pre-B stages`. Depending on the stage of transformation, ALL can be classified into different subtypes, which may influence the prognosis and treatment approach.

### Classes in the Dataset

2. **`[Malignant] Pro-B`**: These are the most `immature form of B cells that have turned malignant`. Pro-B cells are an early stage in B cell development, and their malignancy is a strong indicator of aggressive leukemia. Accurate classification of these cells can be crucial in determining a more intensive treatment approach.

3. **`[Malignant] early Pre-B`**: This class includes `malignant cells that are in the early stage of the Pre-B cell phase`. Pre-B cells are immature B cells, and when these cells become malignant, they proliferate uncontrollably. Early detection of these cells is vital for starting treatment before the cancer advances.

4. **`[Malignant] Pre-B`**: This category involves `malignant cells in a more developed Pre-B stage`. The presence of these cells indicates a progression of the leukemia, and identifying them accurately helps in staging the disease and tailoring the treatment.

**Organize your dataset directory for YOLOv8**

```python


import os
import shutil
from sklearn.model_selection import train_test_split

# Define the paths
original_dataset_dir = 'Blood cell Cancer [ALL]'
base_dir = 'datasets/BloodCellCancer'

# Define new paths for train, val, and test
train_dir = os.path.join(base_dir, 'train')
val_dir = os.path.join(base_dir, 'val')
test_dir = os.path.join(base_dir, 'test')

# Create directories if they do not exist
os.makedirs(train_dir, exist_ok=True)
os.makedirs(val_dir, exist_ok=True)
os.makedirs(test_dir, exist_ok=True)

# List the classes
classes = ['Benign', '[Malignant] early Pre-B', '[Malignant] Pre-B', '[Malignant] Pro-B']

# Function to copy images
def copy_images(class_name, src_dir, train_dst, val_dst, test_dst, train_ratio=0.7, val_ratio=0.15):
    # List all images in the class
    images = os.listdir(src_dir)
    images = [os.path.join(src_dir, img) for img in images if img.endswith('.jpg')]

    # Split into train, val, and test
    train_images, temp_images = train_test_split(images, train_size=train_ratio, random_state=42)
    val_images, test_images = train_test_split(temp_images, test_size=(1 - train_ratio - val_ratio)/(1 - train_ratio), random_state=42)

    # Create class directories in train, val, and test
    train_class_dir = os.path.join(train_dst, class_name.replace('[Malignant] ', 'Malignant '))
    val_class_dir = os.path.join(val_dst, class_name.replace('[Malignant] ', 'Malignant '))
    test_class_dir = os.path.join(test_dst, class_name.replace('[Malignant] ', 'Malignant '))
    os.makedirs(train_class_dir, exist_ok=True)
    os.makedirs(val_class_dir, exist_ok=True)
    os.makedirs(test_class_dir, exist_ok=True)

    # Copy images
    for img in train_images:
        shutil.copy(img, train_class_dir)

    for img in val_images:
        shutil.copy(img, val_class_dir)

    for img in test_images:
        shutil.copy(img, test_class_dir)

    # Return one image from train, val, and test for display
    return train_images[0], val_images[0], test_images[0]

# Dictionaries to store image paths for each class
train_sample_images = {}
val_sample_images = {}
test_sample_images = {}

# Copy images for each class and get a sample image from each class
for class_name in classes:
    src_dir = os.path.join(original_dataset_dir, class_name)
    train_img, val_img, test_img = copy_images(class_name, src_dir, train_dir, val_dir, test_dir)
    train_sample_images[class_name] = train_img
    val_sample_images[class_name] = val_img
    test_sample_images[class_name] = test_img

print("Dataset reorganized successfully")

```

> **NOTE**: `train_batch1.jpg` is not a static file copied from the dataset - it is a snapshot of one real batch *after* the augmentation transforms YOLOv8 applies on the fly during classification training, so it looks different every time you retrain. Three default augmentations are visible here, mixed together in the mosaic:
>
> - **Solid black squares**: `RandomErasing` (`erasing=0.4` by default), a cutout-style regularizer that blanks a random patch so the model can't rely on a single region of the image.
> - **Rotated tiles with black-filled corners**: a rotation operation from `auto_augment='randaugment'`, also on by default for classification.
> - **Images with inverted/unnatural colors (green, yellow, deep red)**: other RandAugment operations (`Solarize`, `Invert`, `Posterize`, `AutoContrast`) that alter color aggressively.
>
> These defaults are tuned for natural-image datasets like ImageNet, not stained microscopy slides, where color carries diagnostic information (nucleus/cytoplasm staining). The color-inverting operations in particular produce training examples that don't resemble any realistic staining or lighting variation. This notebook keeps the defaults as-is for simplicity, but if you were optimizing this model for real use, `auto_augment` and `erasing` would be worth tuning or disabling.

##### Why Recall Matters More Than Accuracy in Blood Cancer Cell Classification

In this dataset, a **``False Negative``** (a malignant cell — `Pro-B`, `early Pre-B`, or `Pre-B` — classified as `Benign`) is usually _``far more costly``_ than a **``False Positive``** (a benign cell flagged for a closer look). This is why, for cancer-screening models like this one, **``Recall (Sensitivity)``** per class is often given more weight than plain **``Accuracy``**: a model can reach a high overall `top1_acc` while still missing a meaningful fraction of malignant cases.

`This is also why per-class metrics matter here`: check the confusion matrix above — specifically, how many malignant images were misclassified as `Benign` — rather than relying only on the single aggregate accuracy number `model.val()` prints.

**Display Results of prediction over test dataset**

---

```python
import os
import random
import matplotlib.pyplot as plt
from PIL import Image

# Define the class folders
class_dirs = {
    'Benign': '/content/datasets/BloodCellCancer/test/Benign',
    'Malignant Early Pre-B': '/content/datasets/BloodCellCancer/test/Malignant early Pre-B',
    'Malignant Pre-B': '/content/datasets/BloodCellCancer/test/Malignant Pre-B',
    'Malignant Pro-B': '/content/datasets/BloodCellCancer/test/Malignant Pro-B'
}

# Randomly select 4 images from each class
random_images = {}

for class_name, dir_path in class_dirs.items():
    # Get all .jpg files in the folder
    images = [os.path.join(dir_path, f) for f in os.listdir(dir_path) if f.endswith('.jpg')]
    # Select 4 random images
    random_images[class_name] = random.sample(images, 4)

# Function to display groundtruth and prediction side by side
def visualize_predictions(images, model):
    # Create a subplot for 4 images
    fig, axs = plt.subplots(4, 2, figsize=(10, 20))  # 4 rows, 2 columns (groundtruth and prediction)

    for i, img_path in enumerate(images):
        # Load the image
        img = Image.open(img_path)

        # Get the groundtruth label from the directory name
        groundtruth = img_path.split('/')[-2]

        # Run prediction on the image
        prediction_result = model(img_path)

        # Display the original image (groundtruth)
        axs[i, 0].imshow(img)
        axs[i, 0].set_title(f"Groundtruth: {groundtruth}")
        axs[i, 0].axis('off')

        # Display the prediction result
        axs[i, 1].imshow(prediction_result[0].plot(show=False))
        axs[i, 1].set_title(f"Prediction")
        axs[i, 1].axis('off')

    # Adjust the layout
    plt.tight_layout()
    plt.show()

# Display predictions and groundtruth for 4 random images of each class
for class_name, images in random_images.items():
    print(f"\nVisualizing for class: {class_name}")
    visualize_predictions(images, model)
```
